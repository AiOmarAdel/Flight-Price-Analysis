```python
# Showing All Airlines And Frequenceis ascending in DataSet # 
Data["airline"].value_counts(ascending=True)
```

```python
# Show Bar Graph of Airlines with Frequenceis in plots #  
plt.figure(figsize=(6,6))
plt.title("Data of Airlines With Frequencies of AirLines DataSet")
plt.xlabel("Name of AirLines")
plt.ylabel("Frequencies of Each Airline")
plt.bar(Data["airline"].unique(),Data["airline"].value_counts(ascending=True).values,color=["lightgreen","lightblue"])
plt.grid()
plt.legend()
plt.show()
```

```python
# Show Bar Graph of Airlines with Frequenceis in plots #  
plt.figure(figsize=(6,6))
plt.title("Data of Airlines With Frequencies of AirLines DataSet")
plt.xlabel("Name of AirLines")
plt.ylabel("Frequencies of Each Airline")
plt.barh(Data["airline"].unique(),Data["airline"].value_counts(ascending=True).values,color=["lightgreen","lightblue"])
plt.grid()
plt.legend()
plt.show()
```

```python
Data.head()
```

```python
# Showing All departure_time And Frequenceis in DataSet # 
data_departure_time=Data["departure_time"].value_counts()
```

```python
# Showing All arrival_time And Frequenceis in DataSet # 
data_arrival_time=Data["arrival_time"].value_counts()
```

```python
# Show Bar Graph of departure_time And arrival_time with Frequenceis in plots #  
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("Distrebution of Departure time ")
plt.xlabel("D Time ")
plt.ylabel("D Frequencey")
plt.bar(data_departure_time.index,data_departure_time.values,color=["lightgreen","lightblue"])
plt.grid()
plt.legend()
plt.xticks(rotation=45)
#---------------------# 
plt.subplot(1,2,2)
plt.title("Distrebution of Arrival time ")
plt.xlabel("AR Time ")
plt.ylabel("AR Frequencey")
plt.bar(data_arrival_time.index,data_arrival_time.values,color=["green","blue"])
plt.grid()
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
#---------------------#
plt.show()
```

```python
# Showing All source_city And Frequenceis in DataSet # 
data_source_city=Data["source_city"].value_counts()
data_source_city
```

```python
# Showing All destination_city And Frequenceis in DataSet # 
data_destination_city=Data["destination_city"].value_counts()
data_destination_city
```

```python
# Show Bar Graph of source_city And destination_city with Frequenceis in plots #  
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("Distrebution of Source city ")
plt.xlabel("Source ")
plt.ylabel("Frequencey")
plt.barh(data_source_city.index,data_source_city.values,color=["red","yellow"])
plt.grid()
plt.legend()
plt.xticks(rotation=45)
#---------------------# 
plt.subplot(1,2,2)
plt.title("Distrebution of Destination city ")
plt.xlabel("Destination ")
plt.ylabel("Frequencey")
plt.barh(data_destination_city.index,data_destination_city.values,color=["orange","red"])
plt.grid()
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
#---------------------#
plt.show()
```

```python
# Groupby The Airlines And Chack Their Mean of The Prices #   
Data.groupby("airline")["price"].mean()
```

```python
# Drowing a Categorical plot Showing the Mean Tickets prices for Each Airlines # 
sns.catplot(data=Data,x="airline",y="price",palette="rocket",kind="bar",hue="class")
plt.show()
```

```python
# Check the Mean Tickets prices board on The Departure time # 
Data.groupby("departure_time")["price"].mean()
```

```python
# Check the Mean Tickets prices board on The Departure time # 
Data.groupby("arrival_time")["price"].mean()
```
